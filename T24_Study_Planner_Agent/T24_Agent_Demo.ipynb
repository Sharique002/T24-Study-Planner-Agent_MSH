## Goal 1 — Add tasks and make a decision
The agent must add two tasks and then build a schedule from the tool results.

```python
answer = agent.run(
    'Add my CSE476 Agentic AI project due 2026-09-15 and my DBMS assignment due 2026-09-20. Then tell me what I should study first.'
)
print('Agent:', answer)
print('\n--- MULTI-STEP TRACE ---')
for item in agent.trace:
    print(item)
```

## Goal 2 — Update the plan
A new deadline is added. The same agent instance is reused, so earlier tasks remain in memory.

```python
answer = agent.run(
    'Also add my AI viva due 2026-09-10 and rebuild my study plan.'
)
print('Agent:', answer)
print('\n--- MULTI-STEP TRACE ---')
for item in agent.trace:
    print(item)
```

## Goal 3 — Prove memory
No deadlines are repeated. The agent must use the tasks remembered from the earlier turns.

```python
answer = agent.run(
    'Which task should I work on first? I already gave you my deadlines earlier.'
)
print('Agent:', answer)
print('\n--- MULTI-STEP TRACE ---')
for item in agent.trace:
    print(item)
```
